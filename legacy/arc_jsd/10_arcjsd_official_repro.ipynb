## Validation protocol

1. Save `official_scores.json` (per-source scores for ≥5 examples).
2. In a FRESH runtime, install the MAIN repo bundle and run
   `attribute --method arc_jsd` with `generation.model_id: Qwen/Qwen2.5-0.5B-Instruct`
   on the same examples.
3. Compare rankings (Spearman) and top-1 agreement. Record the outcome in the main
   repo: PROGRESS.md + (if diverging) FAILURES.md with the observed numbers.
4. Only then may the `experimental` label on the native method be dropped.
